# Deep Learning Facial Verification

This notebook compares two face images and answers: **are these images likely to show the same person?**

It uses OpenCV to find the largest face and a pretrained TensorFlow MobileNetV2 network to create normalized deep image embeddings. For production identity verification, replace the backbone with a face-recognition model trained on facial data and calibrate the threshold on a representative validation set.

```python
from pathlib import Path
import urllib.request

import cv2
import matplotlib.pyplot as plt
import numpy as np
import tensorflow as tf

print(f"TensorFlow: {tf.__version__}")
print(f"OpenCV: {cv2.__version__}")

# ImageNet weights provide a useful baseline embedding without training a model here.
embedding_model = tf.keras.applications.MobileNetV2(
    weights="imagenet",
    include_top=False,
    pooling="avg",
    input_shape=(224, 224, 3),
)

# YuNet is used because some OpenCV 5 Windows builds omit CascadeClassifier.
face_detector = None
if hasattr(cv2, "FaceDetectorYN"):
    detector_path = Path("face_detection_yunet_2023mar.onnx")
    if not detector_path.exists():
        detector_url = (
            "https://github.com/opencv/opencv_zoo/raw/refs/heads/main/models/"
            "face_detection_yunet/face_detection_yunet_2023mar.onnx"
        )
        try:
            urllib.request.urlretrieve(detector_url, detector_path)
        except Exception as error:
            print(f"Could not download the face detector: {error}")
    if detector_path.exists():
        face_detector = cv2.FaceDetectorYN.create(
            str(detector_path), "", (320, 320), 0.8, 0.3, 5000
        )

print(f"Embedding dimensions: {embedding_model.output_shape[-1]}")
print("Face detector: YuNet" if face_detector else "Face detector: full-image fallback (use one centered face per image)")
```

```python
def load_face(image_path, target_size=(224, 224)):
    """Load an image and return the largest detected face, or the full image."""
    image_path = Path(image_path)
    image = cv2.imread(str(image_path))
    if image is None:
        raise FileNotFoundError(f"Could not read image: {image_path}")

    faces = []
    if face_detector is not None:
        height, width = image.shape[:2]
        face_detector.setInputSize((width, height))
        _, detections = face_detector.detect(image)
        if detections is not None:
            faces = [
                tuple(int(value) for value in detection[:4])
                for detection in detections
            ]

    if len(faces) == 0:
        height, width = image.shape[:2]
        face = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        return cv2.resize(face, target_size), (0, 0, width, height)

    x, y, width, height = max(faces, key=lambda box: box[2] * box[3])
    padding_x, padding_y = int(width * 0.15), int(height * 0.15)
    x1, y1 = max(0, x - padding_x), max(0, y - padding_y)
    x2 = min(image.shape[1], x + width + padding_x)
    y2 = min(image.shape[0], y + height + padding_y)
    face = cv2.cvtColor(image[y1:y2, x1:x2], cv2.COLOR_BGR2RGB)
    return cv2.resize(face, target_size), (x, y, width, height)


def face_embedding(image_path):
    face, _ = load_face(image_path)
    batch = tf.keras.applications.mobilenet_v2.preprocess_input(
        face.astype(np.float32)[None, ...]
    )
    embedding = embedding_model.predict(batch, verbose=0)[0]
    return embedding / (np.linalg.norm(embedding) + 1e-8)


def verify_faces(reference_path, candidate_path, threshold=0.72, show=True):
    """Return similarity and a same-person decision for two image paths."""
    reference_face, reference_box = load_face(reference_path)
    candidate_face, candidate_box = load_face(candidate_path)
    reference_embedding = face_embedding(reference_path)
    candidate_embedding = face_embedding(candidate_path)
    similarity = float(np.dot(reference_embedding, candidate_embedding))
    verified = similarity >= threshold

    if show:
        figure, axes = plt.subplots(1, 2, figsize=(8, 4))
        axes[0].imshow(reference_face)
        axes[0].set_title(f"Reference\nface box: {reference_box}")
        axes[1].imshow(candidate_face)
        axes[1].set_title(f"Candidate\nface box: {candidate_box}")
        for axis in axes:
            axis.axis("off")
        figure.suptitle(
            f"{'VERIFIED' if verified else 'NOT VERIFIED'} | cosine similarity: {similarity:.3f}"
        )
        plt.tight_layout()
        plt.show()

    return {"verified": verified, "similarity": similarity, "threshold": threshold}


# Update these paths, then run the final line.
REFERENCE_IMAGE = "reference.jpg"
CANDIDATE_IMAGE = "candidate.jpg"

# result = verify_faces(REFERENCE_IMAGE, CANDIDATE_IMAGE)
# print(result)
```

```python
# Smoke test: validates execution and similarity behavior without real face images.
smoke_reference = np.zeros((224, 224, 3), dtype=np.uint8)
smoke_reference[:, :, 0] = 220
smoke_reference[60:165, 60:165, 1:] = 120
smoke_candidate_same = smoke_reference.copy()
smoke_candidate_different = np.zeros((224, 224, 3), dtype=np.uint8)
smoke_candidate_different[:, :, 2] = 220

cv2.imwrite("smoke_reference.jpg", cv2.cvtColor(smoke_reference, cv2.COLOR_RGB2BGR))
cv2.imwrite("smoke_candidate_same.jpg", cv2.cvtColor(smoke_candidate_same, cv2.COLOR_RGB2BGR))
cv2.imwrite("smoke_candidate_different.jpg", cv2.cvtColor(smoke_candidate_different, cv2.COLOR_RGB2BGR))

same_result = verify_faces("smoke_reference.jpg", "smoke_candidate_same.jpg", show=False)
different_result = verify_faces("smoke_reference.jpg", "smoke_candidate_different.jpg", show=False)
print("Same synthetic image:", same_result)
print("Different synthetic image:", different_result)
assert same_result["similarity"] > 0.99
assert different_result["similarity"] < same_result["similarity"]
```

```python
# Interactive verification: upload two photos, then click Verify.
import ipywidgets as widgets
from IPython.display import display

reference_upload = widgets.FileUpload(accept="image/*", multiple=False, description="Reference photo")
candidate_upload = widgets.FileUpload(accept="image/*", multiple=False, description="Verification photo")
verify_button = widgets.Button(description="Verify", button_style="success", icon="check")
verification_output = widgets.Output()


def uploaded_file(upload_widget):
    value = upload_widget.value
    if not value:
        return None
    file_info = next(iter(value.values())) if isinstance(value, dict) else value[0]
    file_name = file_info["name"]
    file_bytes = file_info["content"]
    Path(file_name).write_bytes(bytes(file_bytes))
    return file_name


def on_verify_clicked(_):
    with verification_output:
        verification_output.clear_output()
        reference_path = uploaded_file(reference_upload)
        candidate_path = uploaded_file(candidate_upload)
        if reference_path is None or candidate_path is None:
            print("Upload both photos first.")
            return
        try:
            result = verify_faces(reference_path, candidate_path)
            print(result)
        except (FileNotFoundError, ValueError) as error:
            print(f"Verification failed: {error}")


verify_button.on_click(on_verify_clicked)
display(reference_upload, candidate_upload, verify_button, verification_output)
```

```python
# Face unlock: capture a webcam enrollment set, then verify future webcam frames.
def capture_webcam_image(output_path="webcam_candidate.jpg", camera_index=0, countdown_seconds=5):
    import time

    camera = cv2.VideoCapture(camera_index, cv2.CAP_DSHOW)
    if not camera.isOpened():
        camera.release()
        camera = cv2.VideoCapture(camera_index)
    if not camera.isOpened():
        camera.release()
        raise RuntimeError("Could not open the webcam. Check Windows camera permissions.")

    print(f"Look at the camera; capturing automatically in {countdown_seconds} seconds.")
    captured = False
    started_at = time.monotonic()
    window_name = "Face unlock - automatic capture"
    try:
        cv2.namedWindow(window_name, cv2.WINDOW_NORMAL)
        while True:
            success, frame = camera.read()
            if not success:
                raise RuntimeError("The webcam opened but did not provide a frame.")
            seconds_left = max(0, countdown_seconds - int(time.monotonic() - started_at))
            preview = frame.copy()
            cv2.putText(preview, f"Capture in {seconds_left}s | Space: now | Q/Esc: cancel", (20, 35), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2, cv2.LINE_AA)
            cv2.imshow(window_name, preview)
            key = cv2.waitKey(30) & 0xFF
            if key == 32 or time.monotonic() - started_at >= countdown_seconds:
                captured = bool(cv2.imwrite(output_path, frame))
                break
            if key in (ord("q"), 27):
                break
    finally:
        camera.release()
        cv2.destroyAllWindows()
    if not captured:
        raise RuntimeError("Webcam capture cancelled or could not save the image.")
    return output_path


def capture_enrollment_images(output_dir="enrollment_images", image_count=20, interval_seconds=1.0, camera_index=0):
    """Capture multiple webcam images for a more stable enrollment embedding."""
    import time

    output_dir = Path(output_dir)
    output_dir.mkdir(exist_ok=True)
    camera = cv2.VideoCapture(camera_index, cv2.CAP_DSHOW)
    if not camera.isOpened():
        camera.release()
        camera = cv2.VideoCapture(camera_index)
    if not camera.isOpened():
        camera.release()
        raise RuntimeError("Could not open the webcam. Check Windows camera permissions.")

    image_paths = []
    started_at = time.monotonic()
    next_capture_at = 0.0
    window_name = "Face enrollment - look at camera"
    print(f"Capturing {image_count} enrollment images over about {image_count * interval_seconds:.0f} seconds.")
    print("Stay in good light and slowly turn your head a little. Press Q or Esc to cancel.")
    try:
        cv2.namedWindow(window_name, cv2.WINDOW_NORMAL)
        while len(image_paths) < image_count:
            success, frame = camera.read()
            if not success:
                raise RuntimeError("The webcam opened but did not provide a frame.")
            elapsed = time.monotonic() - started_at
            remaining = image_count - len(image_paths)
            preview = frame.copy()
            cv2.putText(preview, f"Enrollment: {len(image_paths)}/{image_count} | next in {max(0, next_capture_at - elapsed):.1f}s", (20, 35), cv2.FONT_HERSHEY_SIMPLEX, 0.65, (0, 255, 0), 2, cv2.LINE_AA)
            cv2.imshow(window_name, preview)
            key = cv2.waitKey(30) & 0xFF
            if key in (ord("q"), 27):
                break
            if elapsed >= next_capture_at:
                image_path = output_dir / f"enrollment_{len(image_paths) + 1:02d}.jpg"
                if cv2.imwrite(str(image_path), frame):
                    image_paths.append(str(image_path))
                    next_capture_at += interval_seconds
    finally:
        camera.release()
        cv2.destroyAllWindows()

    if len(image_paths) != image_count:
        raise RuntimeError(f"Enrollment cancelled after {len(image_paths)} of {image_count} images.")
    print(f"Saved {len(image_paths)} enrollment images in {output_dir}")
    return image_paths


def create_enrollment_from_webcam(image_count=20, output_dir="enrollment_images", enrollment_file="enrollment_embedding.npy"):
    """Capture images and save their averaged normalized embedding."""
    image_paths = capture_enrollment_images(output_dir=output_dir, image_count=image_count)
    embeddings = np.stack([face_embedding(image_path) for image_path in image_paths])
    enrollment_embedding = embeddings.mean(axis=0)
    enrollment_embedding /= np.linalg.norm(enrollment_embedding) + 1e-8
    np.save(enrollment_file, enrollment_embedding)
    print(f"Saved averaged enrollment embedding: {enrollment_file}")
    return {"images": image_paths, "embedding_file": enrollment_file}


def enroll_face(reference_path=None, enrollment_file="enrollment_embedding.npy"):
    """Load the averaged enrollment embedding, or create one from a reference photo."""
    enrollment_path = Path(enrollment_file)
    if reference_path is None and enrollment_path.exists():
        return np.load(enrollment_path)
    reference_path = Path(reference_path or REFERENCE_IMAGE)
    if not reference_path.exists():
        raise FileNotFoundError(f"Enrollment photo not found: {reference_path}")
    return face_embedding(reference_path)


def unlock_with_webcam(reference_path=None, threshold=0.85, camera_index=0, countdown_seconds=5):
    """Capture one webcam frame and compare it with the averaged enrollment."""
    enrollment_embedding = enroll_face(reference_path)
    webcam_path = capture_webcam_image(camera_index=camera_index, countdown_seconds=countdown_seconds)
    captured_embedding = face_embedding(webcam_path)
    similarity = float(np.dot(enrollment_embedding, captured_embedding))
    unlocked = similarity >= threshold
    status = "UNLOCKED" if unlocked else "LOCKED"
    print(f"{status} | cosine similarity: {similarity:.3f} | threshold: {threshold:.3f}")
    return {"unlocked": unlocked, "similarity": similarity, "threshold": threshold, "image": webcam_path}


REFERENCE_IMAGE = "reference..jpg"
ENROLLMENT_FILE = "enrollment_embedding.npy"
```

```python
# Live enrollment preview: the separate camera window shows every frame before capture.
def capture_enrollment_images(output_dir="enrollment_images", image_count=20, interval_seconds=1.0, camera_index=0):
    import time

    output_dir = Path(output_dir)
    output_dir.mkdir(exist_ok=True)
    camera = cv2.VideoCapture(camera_index, cv2.CAP_DSHOW)
    if not camera.isOpened():
        camera.release()
        camera = cv2.VideoCapture(camera_index)
    if not camera.isOpened():
        camera.release()
        raise RuntimeError("Could not open the webcam. Check Windows camera permissions.")

    image_paths = []
    started_at = time.monotonic()
    next_capture_at = 0.0
    window_name = "Face enrollment - live preview"
    print("A live camera window is open. Check your face framing there while images are captured.")
    print("Stay visible in the green box. Press Q or Esc in that window to cancel.")
    try:
        cv2.namedWindow(window_name, cv2.WINDOW_NORMAL)
        cv2.resizeWindow(window_name, 960, 720)
        while len(image_paths) < image_count:
            success, frame = camera.read()
            if not success:
                raise RuntimeError("The webcam opened but did not provide a frame.")

            elapsed = time.monotonic() - started_at
            preview = frame.copy()
            face_box = None
            if face_detector is not None:
                height, width = frame.shape[:2]
                face_detector.setInputSize((width, height))
                _, detections = face_detector.detect(frame)
                if detections is not None and len(detections) > 0:
                    detection = max(detections, key=lambda item: item[2] * item[3])
                    face_box = tuple(int(value) for value in detection[:4])
                    x, y, box_width, box_height = face_box
                    cv2.rectangle(preview, (x, y), (x + box_width, y + box_height), (0, 255, 0), 3)

            status = f"Images: {len(image_paths)}/{image_count}"
            if face_box is None:
                status += " | Face not detected - move closer / add light"
            else:
                status += " | Face detected"
            cv2.putText(preview, status, (20, 40), cv2.FONT_HERSHEY_SIMPLEX, 0.75, (0, 255, 0), 2, cv2.LINE_AA)
            cv2.putText(preview, "Look at camera | Q/Esc: cancel", (20, 75), cv2.FONT_HERSHEY_SIMPLEX, 0.65, (255, 255, 255), 2, cv2.LINE_AA)
            cv2.imshow(window_name, preview)
            key = cv2.waitKey(30) & 0xFF
            if key in (ord("q"), 27):
                break
            if face_box is not None and elapsed >= next_capture_at:
                image_path = output_dir / f"enrollment_{len(image_paths) + 1:02d}.jpg"
                if cv2.imwrite(str(image_path), frame):
                    image_paths.append(str(image_path))
                    next_capture_at += interval_seconds
    finally:
        camera.release()
        cv2.destroyAllWindows()

    if len(image_paths) != image_count:
        raise RuntimeError(f"Enrollment cancelled after {len(image_paths)} of {image_count} images.")
    print(f"Saved {len(image_paths)} enrollment images in {output_dir}")
    return image_paths
```
